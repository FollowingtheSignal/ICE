```python
# Load the Deportation Data Project facility-location dataset
facilities = pd.read_excel(
    "~/Downloads/detention-facilities_filtered.xlsx",
    sheet_name="data"
)

# Start with the original county values
# Run this on the dataset before any county imputation
ice["county_original"] = ice["county"].copy()

# Track how each county was obtained
ice["county_source"] = np.where(
    ice["county"].notna(),
    "Original",
    "Unresolved"
)

print("Missing counties before:", ice["county"].isna().sum())


# ---------------------------------------------------
# METHOD 1: Match counties using detention facility codes
# ---------------------------------------------------

# Check for conflicting county assignments
facility_conflicts = (
    facilities.dropna(subset=["detention_facility_code", "county"])
    .groupby("detention_facility_code")["county"]
    .nunique()
)

print(
    "Facility codes with conflicting counties:",
    (facility_conflicts > 1).sum()
)

# Use facility codes associated with exactly one county
valid_codes = facility_conflicts[facility_conflicts == 1].index

county_lookup = (
    facilities[
        facilities["detention_facility_code"].isin(valid_codes)
    ]
    .dropna(subset=["county"])
    .drop_duplicates(subset=["detention_facility_code"])
    .set_index("detention_facility_code")["county"]
)

# Find matching counties
matched_county = ice["detention_facility_code"].map(county_lookup)

# Identify records that can be filled
facility_mask = (
    ice["county"].isna() &
    matched_county.notna()
)

# Fill missing values
ice.loc[facility_mask, "county"] = matched_county[facility_mask]

# Record the source
ice.loc[facility_mask, "county_source"] = "Facility code"

print("Filled using facility codes:", facility_mask.sum())


# ---------------------------------------------------
# METHOD 2: Match counties using city and state
# ---------------------------------------------------

# Use original records with known city, state, and county
known = ice.dropna(
    subset=["city", "state", "county_original"]
).copy()

# Count different counties associated with each city-state pair
county_counts = (
    known.groupby(["city", "state"])["county_original"]
    .nunique()
)

# Exclude city-state combinations with multiple known counties
valid_locations = county_counts[county_counts == 1].index

# Create a city-state-to-county lookup
county_mapping = (
    known.groupby(["city", "state"])["county_original"]
    .first()
    .loc[valid_locations]
)

# Match the city and state of every record
location_keys = pd.MultiIndex.from_frame(
    ice[["city", "state"]]
)

inferred_counties = pd.Series(
    location_keys.map(county_mapping),
    index=ice.index
)

# Identify remaining missing counties that can be filled
city_mask = (
    ice["county"].isna() &
    inferred_counties.notna()
)

# Fill missing values
ice.loc[city_mask, "county"] = inferred_counties[city_mask]

# Record the source
ice.loc[city_mask, "county_source"] = "City-state inference"

print("Filled using city-state matching:", city_mask.sum())


# ---------------------------------------------------
# FINAL SUMMARY
# ---------------------------------------------------

print("\nCounty Imputation Summary")
print("-" * 35)

print("Total records:", len(ice))
print("Missing before:", ice["county_original"].isna().sum())
print("Filled using facility code:", facility_mask.sum())
print("Filled using city-state:", city_mask.sum())
print("Total filled:", (facility_mask | city_mask).sum())
print("Still missing:", ice["county"].isna().sum())

print("\nCounty value sources:")
print(ice["county_source"].value_counts())
```

```python
import unicodedata

def normalize_county_for_comparison(value):
    if pd.isna(value):
        return None

    value = str(value).strip().upper()

    # Remove accents
    value = "".join(
        c for c in unicodedata.normalize("NFKD", value)
        if not unicodedata.combining(c)
    )

    # Remove common suffixes
    for suffix in [
        " COUNTY",
        " CITY",
        " CITY AND BOROUGH OF"
    ]:
        if value.endswith(suffix):
            value = value.removesuffix(suffix)

    return value.strip()

comparison["ice_normalized"] = (
    comparison["county_original"].map(normalize_county_for_comparison)
)

comparison["vera_normalized"] = (
    comparison["vera_county"].map(normalize_county_for_comparison
                                 )
)

comparison["normalized_match"] = (
    comparison["ice_normalized"] ==
    comparison["vera_normalized"]
)

print(
    "Agreement after normalization:",
    comparison["normalized_match"].mean().round(4)
)

print(
    "Remaining disagreements:",
    (~comparison["normalized_match"]).sum()
)
```

```python
# Facility codes with any remaining disagreement
conflicting_codes = set(
    comparison.loc[
        ~comparison["normalized_match"],
        "detention_facility_code"
    ]
)

# Exclude them from Vera's lookup
safe_lookup = vera_lookup.loc[
    ~vera_lookup.index.isin(conflicting_codes)
]

# Check how many missing counties we can fill
matched_safe = ice["detention_facility_code"].map(safe_lookup)

fillable = (
    ice["county"].isna() &
    matched_safe.notna()
)

print("Conflicting facility codes excluded:", len(conflicting_codes))
print("Missing county records potentially fillable:", fillable.sum())
print("Remaining if filled:", ice["county"].isna().sum() - fillable.sum())
```

```python
census["county_fips"] = (
    census["STATEFP"] + census["COUNTYFP"]
)
census.head()
```

```python
import unicodedata
import re

def normalize_county_for_fips(name):
    if pd.isna(name):
        return None

    name = str(name).upper().strip()

    # Remove accents (Doña Ana -> Dona Ana)
    name = "".join(
        c for c in unicodedata.normalize("NFKD", name)
        if not unicodedata.combining(c)
    )

    # Remove common geographic suffixes
    name = re.sub(
        r"\s+(COUNTY|PARISH|BOROUGH|CENSUS AREA)$",
        "",
        name
    )

    return name.strip()

ice["county_match"] = ice["county"].map(normalize_county_for_fips)
census["county_match"] = census["COUNTYNAME"].map(normalize_county_for_fips)
```

## 7. Aggregate ICE Detention Records by Month:
This section summarizes the detention records by month to examine changes over time.

First, I group the records by their book-in month and calculate the total number of detention stints, the number of unique individuals, and the median length of facility stays. I also count records in each criminality and conviction category, including missing values.

Next, I calculate the percentage of monthly detention records classified as convicted criminals, pending criminal charges, or other immigration violators.

To examine violent and nonviolent convictions, I filter the dataset to include only records classified as convicted criminals. I then calculate the percentage of violent, nonviolent, and unclassified convictions within this group.

Finally, I merge these results into the monthly summary DataFrame, creating one table containing the statistics used in the analysis and visualizations.


```python
monthly = (
    ice_analysis
    .assign(month=lambda df: df["book_in_date_time"].dt.to_period("M"))
    .groupby("month")
    .agg(
        records=("detention_facility_code", "size"),
        unique_individuals=("unique_identifier", "nunique"),
        median_stint_days=("stint_length_days", "median"),
        violent = ("conviction", lambda x: (x=='Violent crime').sum()),
        non_violent = ("conviction", lambda x: (x=='Nonviolent crime').sum()),
        conviction_missing=("conviction", lambda x: x.isna().sum()),
        convicted_criminal = ("book_in_criminality", lambda x: (x=='1 Convicted Criminal').sum()),
        pending_criminal_charges = ("book_in_criminality", lambda x: (x=='2 Pending Criminal Charges').sum()),
        immigration_violator = ("book_in_criminality",  lambda x: (x=='3 Other Immigration Violator').sum()),
        crime_missing = ("book_in_criminality", lambda x: x.isna().sum())
        
        
    )
    .reset_index()
    .sort_values("month")
)
monthly["median_stint_days"] = monthly["median_stint_days"].round(1)
monthly.head(10)
```

```python
monthly["convicted_pct"] = (
    100 * monthly["convicted_criminal"] / monthly["records"]
).round(2)

monthly["pending_pct"] = (
    100 * monthly["pending_criminal_charges"] / monthly["records"]
).round(2)

monthly["immigration_pct"] = (
    100 * monthly["immigration_violator"] / monthly["records"]
).round(2)
convicted = ice_analysis.loc[
    ice_analysis["book_in_criminality"] == "1 Convicted Criminal"
].copy()
conviction_monthly = (
    convicted
    .assign(month=lambda df: df["book_in_date_time"].dt.to_period("M"))
    .groupby("month")
    .agg(
        convicted_records=("conviction", "size"),
        violent=("conviction", lambda x: (x == "Violent crime").sum()),
        non_violent=("conviction", lambda x: (x == "Nonviolent crime").sum()),
        unclassified=("conviction", lambda x: x.isna().sum())
    )
    .reset_index()
)

conviction_monthly["violent_pct"] = (
    100 * conviction_monthly["violent"]
    / conviction_monthly["convicted_records"]
).round(2)

conviction_monthly["non_violent_pct"] = (
    100 * conviction_monthly["non_violent"]
    / conviction_monthly["convicted_records"]
).round(2)

conviction_monthly["unclassified_pct"] = (
    100 * conviction_monthly["unclassified"]
    / conviction_monthly["convicted_records"]
).round(2)

conviction_extra = conviction_monthly[[
    "month",
    "violent_pct",
    "non_violent_pct",
    "unclassified",
    "unclassified_pct"
]].rename(columns={
    "violent_pct": "violent_among_convicted_pct",
    "non_violent_pct": "nonviolent_among_convicted_pct",
    "unclassified": "conviction_unclassified",
    "unclassified_pct": "unclassified_case_pct"
})

# Remove previously merged columns, if they exist
monthly = monthly.drop(
    columns=conviction_extra.columns.drop("month"),
    errors="ignore"
)

# Merge updated values
monthly = monthly.merge(
    conviction_extra,
    on="month",
    how="left",
    validate="one_to_one"
)

monthly.head()
```

## 8. Calculate Monthly Detention Stints per Individual

To examine whether individuals appeared in more detention-facility records over time, I calculated the average number of stints per person for each month.

First, I excluded records with missing individual identifiers and grouped the remaining records by their book-in month. For each month, I counted the total number of detention records and the number of distinct individuals.

I then divided the number of records by the number of individuals to calculate the average number of detention stints per person. Finally, I exported the results to monthly_stints.csv.

An increase in this measure may indicate more frequent transfers between facilities, repeated detention episodes, or both. It does not directly measure transfers.

```python
df = ice_analysis.dropna(subset=["unique_identifier"]).copy()

# Create year-month from the original booking date
df["year_month"] = (
    pd.to_datetime(df["book_in_date_time"])
      .dt.to_period("M")
      .dt.to_timestamp()
)

monthly_transfers = (
    df.groupby("year_month")
      .agg(
          records=("unique_identifier", "size"),
          individuals=("unique_identifier", "nunique")
      )
      .reset_index()
)

monthly_transfers["stints_per_person"] = (
    monthly_transfers["records"] /
    monthly_transfers["individuals"]
)
print(
    monthly_transfers[
        ["year_month", "records", "individuals", "stints_per_person"]
    ].round(2).to_string(index=False)
)
monthly_transfers.to_csv('monthly_stints.csv')
```
